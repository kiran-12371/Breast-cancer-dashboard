```python
Q1 = df[["Survival Months","Tumor Size"]].quantile(0.25)
Q3 = df[["Survival Months","Tumor Size"]].quantile(0.75)


IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df[["Survival Months","Tumor Size"]] < lower_bound) |
    (df[["Survival Months","Tumor Size"]] > upper_bound)
]

outliers.shape
```

#Bivariate Analysis

```python

pd.crosstab(df['Age'], df['Status'])
```

```python

sns.countplot(
    data=df,
    x='Age',
    hue='Status'
)

plt.title('Age  vs Patient Status')
plt.xticks(rotation=90)
plt.show()
```

```python

sns.scatterplot(
    data=df,
    x='Tumor Size',
    y='Survival Months',
    hue='Status'
)

plt.title('Tumor Size vs Survival Months')
plt.show()
```

```python

pd.crosstab(
    df['Estrogen Status'],
    df['Status']
)
```

```python

sns.countplot(
    data=df,
    x='Estrogen Status',
    hue='Status'
)

plt.title('Estrogen Status vs Patient Status')
plt.show()
```

#Multivariate Analysis

```python

numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python

sns.scatterplot(
    data=df,
    x='Age',
    y='Survival Months',
    size='Tumor Size',
    hue='Status',
      alpha=0.7
)

plt.title('Age, Tumor Size and Survival')
plt.show()
```

```python

sns.boxplot(
    data=df,
    x='Age',
    y='Survival Months',
    hue='Status'
)

plt.title('Survival by Age Group and Status')
plt.xticks(rotation=90)
plt.show()
```

```python

df.groupby('Status',
           observed=False)['Survival Months'].mean()
```

```python

sns.boxplot(
    data=df,
    x='Status',
    y='Survival Months'
)

plt.title('Survival Duration by Patient Status')
plt.show()
```

```python

df['Status'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    title='Patient Status Percentage'
)


plt.show()
```

```python

print("BREAST CANCER DATA INSIGHTS")

print("\n1. Total Patients:")
print(len(df))

print("\n2. Most Common Age Group:")
print(df['Age'].value_counts().idxmax())

print("\n3. Patient Status:")
print(df['Status'].value_counts())

print("\n4. Average Tumor Size:")
print(df['Tumor Size'].mean())

print("\n5. Average Survival Months:")
print(df['Survival Months'].mean())

print("\n6. Highest Death Count Age Group:")
print(
    df[df['Status'] == 'Dead']
    ['Age'].value_counts().idxmax()
)

print("\n7. Average Tumor Size by Status:")
print(df.groupby('Status')['Tumor Size'].mean())
```
